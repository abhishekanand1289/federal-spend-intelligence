```python
df = df[df['federal_action_obligation'] > 0]
```

```python
top_5_agencies = ["General Services Administration", "Department of Defense", "Department of Justice", 
                    "Department of Veterans Affairs", "Department of Homeland Security"]
```

```python
df = df.dropna(subset = 'recipient_name')
```
