```python
## AGENCY TABLE

agency_cols = ['awarding_agency_name', 'awarding_sub_agency_name', 'funding_agency_name']
```

```python
agency_df = df[agency_cols].drop_duplicates()
```

```python
agency_df
```

```python
agency_map = {}
```

```python
for _, row in agency_df.iterrows():
    cursor.execute(""" INSERT INTO agencies(awarding_agency_name, awarding_sub_agency_name, funding_agency_name)
                        VALUES(%s, %s, %s)""",
                  (row['awarding_agency_name'],
                  row['awarding_sub_agency_name'],
                  row['funding_agency_name']))
    agency_id = cursor.lastrowid
    agency_map[(row['awarding_agency_name'],
                row['awarding_sub_agency_name'],
                row['funding_agency_name'])] = agency_id
```

```python
conn.commit()
```

```python
len(agency_map)
```

```python
## RECIPIENTS

recipients_cols = ["recipient_uei", "recipient_name", "recipient_city_name", "recipient_state_name", "recipient_country_name", 
                   "veteran_owned_business", "woman_owned_business", "minority_owned_business", "small_disadvantaged_business"]
```

```python
recipients_df = df[recipients_cols].drop_duplicates()
```

```python
recipients_df
```

```python
recipients_map = {}

for _, row in recipients_df.iterrows():
    cursor.execute("""INSERT INTO recipients(recipient_uei, recipient_name, recipient_city_name, recipient_state_name, 
    recipient_country_name, veteran_owned_business, woman_owned_business, minority_owned_business, small_disadvantaged_business)
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)""",
    (row['recipient_uei'] if pd.notna(row['recipient_uei']) else None,
     row['recipient_name'] if pd.notna(row['recipient_name']) else None,
     row['recipient_city_name'] if pd.notna(row['recipient_city_name']) else None,
     row['recipient_state_name'] if pd.notna(row['recipient_state_name']) else None,
     row['recipient_country_name'] if pd.notna(row['recipient_country_name']) else None,
     row['veteran_owned_business'] if pd.notna(row['veteran_owned_business']) else None,
     row['woman_owned_business'] if pd.notna(row['woman_owned_business']) else None,
     row['minority_owned_business'] if pd.notna(row['minority_owned_business']) else None,
     row['small_disadvantaged_business'] if pd.notna(row['small_disadvantaged_business']) else None))


    recipients_map[row['recipient_uei']] = cursor.lastrowid
```

```python
conn.commit()
```

```python
## LOCATIONS

recipient_locs = df[['recipient_city_name', 'recipient_state_name', 'recipient_country_name']]
recipient_locs.columns = ['city_name', 'state_name', 'country_name']
recipient_locs['location_type'] = 'recipient'
```

```python
recipient_locs
```

```python
performance_locs = df[['primary_place_of_performance_city_name','primary_place_of_performance_state_name',
                       'primary_place_of_performance_country_name']]
performance_locs.columns = ['city_name', 'state_name', 'country_name']
performance_locs['location_type'] = 'performance'
```

```python
performance_locs
```

```python
locs_df = pd.concat([recipient_locs, performance_locs]).drop_duplicates()
```

```python
locs_df
```

```python
locs_map = {}

for _, row in locs_df.iterrows():
    cursor.execute("""INSERT INTO locations(city_name, state_name, country_name, location_type)
    VALUES(%s, %s, %s, %s)""",(
        row['city_name'] if pd.notna(row['city_name']) else None,
        row['state_name'] if pd.notna(row['state_name']) else None,
        row['country_name'] if pd.notna(row['country_name']) else None,
        row['location_type']
    ))

    locs_map[(row['city_name'],
             row['state_name'],
             row['country_name'],
             row['location_type'])] = cursor.lastrowid
```

```python
conn.commit()
```

```python
df['action_date'] = pd.to_datetime(df['action_date'], errors = 'coerce').dt.date
df['period_of_performance_start_date'] = pd.to_datetime(df['period_of_performance_start_date'], errors = 'coerce').dt.date
df['period_of_performance_current_end_date'] = pd.to_datetime(df['period_of_performance_current_end_date'], errors = 'coerce').dt.date
```

```python
## CONTRACTS

def get_agency_id(agency, sub_agency, funding):
    key = (agency, sub_agency, funding)
    return agency_map.get(key)

def get_recipient_id(uei):
    if pd.isna(uei):
        return None
    return recipients_map.get(uei)

def get_location_id(city, state, country, loc_type):
    key = (city if pd.notna(city) else None,
          state if pd.notna(state) else None,
          country if pd.notna(country) else None,
          loc_type)
    return locs_map.get(key)
```

```python
batch = []
batch_size = 1000

for index, row in df.iterrows():
    batch.append((row['contract_award_unique_key'] if pd.notna(row['contract_award_unique_key']) else None,
        row['award_id_piid'] if pd.notna(row['award_id_piid']) else None,
        get_agency_id(row['awarding_agency_name'], row['awarding_sub_agency_name'], row['funding_agency_name']),
        get_recipient_id(row['recipient_uei']),
        get_location_id(row['recipient_city_name'], row['recipient_state_name'], row['recipient_country_name'], 'recipient'),
        get_location_id(row['primary_place_of_performance_city_name'], row['primary_place_of_performance_state_name'], row['primary_place_of_performance_country_name'], 'performance'),
        float(row['federal_action_obligation']) if pd.notna(row['federal_action_obligation']) else None,
        float(row['total_dollars_obligated']) if pd.notna(row['total_dollars_obligated']) else None,
        row['action_date'],
        int(row['action_date_fiscal_year']) if pd.notna(row['action_date_fiscal_year']) else None,
        row['period_of_performance_start_date'],
        row['period_of_performance_current_end_date'],
        row['transaction_description'] if pd.notna(row['transaction_description']) else None,
        row['award_type'] if pd.notna(row['award_type']) else None,
        row['type_of_contract_pricing'] if pd.notna(row['type_of_contract_pricing']) else None,
        row['extent_competed'] if pd.notna(row['extent_competed']) else None,
        int(row['number_of_offers_received']) if pd.notna(row['number_of_offers_received']) else None,
        row['naics_code'] if pd.notna(row['naics_code']) else None,
        row['naics_description'] if pd.notna(row['naics_description']) else None,
        row['product_or_service_code'] if pd.notna(row['product_or_service_code']) else None,
        row['product_or_service_code_description'] if pd.notna(row['product_or_service_code_description']) else None,
    ))

    if len(batch) == batch_size:
        cursor.executemany("""
        INSERT INTO contracts(contract_award_unique_key, award_id_piid, agency_id, recipient_id, recipient_location_id, 
                                performance_location_id, federal_action_obligation, total_dollars_obligated,
                                action_date, action_date_fiscal_year, period_of_performance_start_date, 
                                period_of_performance_current_end_date, transaction_description, award_type, 
                                type_of_contract_pricing, extent_competed, number_of_offers_received, 
                                naics_code, naics_description, 
                                product_or_service_code, product_or_service_code_description)
        
        VALUE(%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)""", batch)
        conn.commit()
        batch = []
if batch:
    cursor.executemany("""
        INSERT INTO contracts (
            contract_award_unique_key, award_id_piid,
            agency_id, recipient_id,
            recipient_location_id, performance_location_id,
            federal_action_obligation, total_dollars_obligated,
            action_date, action_date_fiscal_year,
            period_of_performance_start_date, period_of_performance_current_end_date,
            transaction_description, award_type,
            type_of_contract_pricing, extent_competed,
            number_of_offers_received, naics_code,
            naics_description, product_or_service_code,
            product_or_service_code_description
        ) VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)
    """, batch)
    conn.commit()

```
